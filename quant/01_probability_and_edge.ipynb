## 1 · Condition on what you observed
A detector flags 80% of jump sessions and 20% of quiet sessions. Only 10% of sessions contain a jump. You observe a flag. What probability should you use for a jump?

Imagine 1,000 sessions. There are 80 flagged jumps and 180 flagged quiet sessions. Among the 260 flags, the jump share is 80/260.

```python
def precision(prior, sensitivity, false_alarm):
        flagged_jumps = prior * sensitivity
        flagged_quiet = (1 - prior) * false_alarm
        return flagged_jumps / (flagged_jumps + flagged_quiet)

p = precision(0.10, 0.80, 0.20)
print(f"Jump probability after a flag: {p:.1%}")
```

## 2 · Turn a probability into a price
A contract pays 10 dollars if a jump occurs and zero otherwise. It costs 3 dollars plus a 20-cent fee. On flagged days, expected net profit is the conditional expected payout minus both costs. Predict its sign before running the cell.

```python
payout, price, fee = 10, 3, 0.20
expected_net = p * payout - price - fee
print(f"Expected net per trade: ${expected_net:.3f}")
print(f"Break-even probability: {(price + fee) / payout:.1%}")
```

## 3 · Test a change in market regime
Hold the detector's two conditional rates fixed. If jumps grow rarer, predict what happens to precision and expected profit. Try changing the false-alarm rate as a second experiment.

```python
for prior in [0.05, 0.10, 0.20]:
        chance = precision(prior, 0.80, 0.20)
        net = chance * payout - price - fee
        print(f"Prior {prior:.0%} | precision {chance:.1%} | net ${net:.2f}")
```
